```python
import os
import shutil
from pathlib import Path

SRC = "/content/football_players"
DST = "/content/football_players_3class"

# Original -> new class IDs
class_map = {
    1: 0,   # goalkeeper
    2: 1,   # player
    3: 2    # referee
}

splits = ["train", "valid", "test"]

# Start fresh
if os.path.exists(DST):
    shutil.rmtree(DST)

for split in splits:

    src_images = Path(SRC) / split / "images"
    src_labels = Path(SRC) / split / "labels"

    dst_images = Path(DST) / split / "images"
    dst_labels = Path(DST) / split / "labels"

    dst_images.mkdir(parents=True, exist_ok=True)
    dst_labels.mkdir(parents=True, exist_ok=True)

    # Copy images
    for img in src_images.iterdir():
        if img.suffix.lower() in [".jpg", ".jpeg", ".png"]:
            shutil.copy2(img, dst_images / img.name)

    # Filter + remap labels
    for label_file in src_labels.glob("*.txt"):

        new_lines = []

        with open(label_file, "r") as f:
            for line in f:

                parts = line.strip().split()

                if len(parts) < 5:
                    continue

                old_class = int(parts[0])

                # Ignore ball
                if old_class == 0:
                    continue

                # Remap remaining classes
                if old_class in class_map:
                    parts[0] = str(class_map[old_class])
                    new_lines.append(" ".join(parts))

        with open(dst_labels / label_file.name, "w") as f:
            f.write("\n".join(new_lines))

print("3-class dataset created:", DST)
```

```python
import os
import random
import cv2
import matplotlib.pyplot as plt

BASE = "/content/football_players_3class"
IMG_DIR = os.path.join(BASE, "train", "images")
LBL_DIR = os.path.join(BASE, "train", "labels")

class_names = {
    0: "goalkeeper",
    1: "player",
    2: "referee"
}

images = [
    f for f in os.listdir(IMG_DIR)
    if f.lower().endswith((".jpg", ".jpeg", ".png"))
]

sample = random.sample(images, min(4, len(images)))

plt.figure(figsize=(16, 12))

for i, img_name in enumerate(sample):
    img_path = os.path.join(IMG_DIR, img_name)
    label_path = os.path.join(
        LBL_DIR,
        os.path.splitext(img_name)[0] + ".txt"
    )

    img = cv2.imread(img_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    h, w = img.shape[:2]

    if os.path.exists(label_path):
        with open(label_path, "r") as f:
            for line in f:
                parts = line.strip().split()

                if len(parts) < 5:
                    continue

                cls = int(parts[0])
                xc, yc, bw, bh = map(float, parts[1:5])

                x1 = int((xc - bw / 2) * w)
                y1 = int((yc - bh / 2) * h)
                x2 = int((xc + bw / 2) * w)
                y2 = int((yc + bh / 2) * h)

                cv2.rectangle(img, (x1, y1), (x2, y2), (255, 0, 0), 2)

                cv2.putText(
                    img,
                    class_names[cls],
                    (x1, max(20, y1 - 5)),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.6,
                    (255, 0, 0),
                    2
                )

    plt.subplot(2, 2, i + 1)
    plt.imshow(img)
    plt.title(img_name)
    plt.axis("off")

plt.tight_layout()
plt.show()
```
